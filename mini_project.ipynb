```python
df['shipping_duration'] = (df['ship_date'] - df['order_date']).dt.days
```

```python
avg_shipping_days = df['shipping_duration'].mean()
```

```python
def classify_shipping_performance(duration, mean_duration):
    if duration < (mean_duration - 1):
        return 'Early'
    elif (mean_duration - 1) <= duration <= (mean_duration + 1):
        return 'On Time / Normal'
    else:
        return 'Delayed'
```

```python
df['shipping_status'] = df['shipping_duration'].apply(
    lambda x: classify_shipping_performance(x, avg_shipping_days)
)
```

```python
shipmode_perf = df.groupby('ship_mode').agg(
    avg_days=('shipping_duration', 'mean'),
    delayed_orders=('shipping_status', lambda x: (x == 'Delayed').sum()),
    total_orders=('order_id', 'count')
)
shipmode_perf['delay_rate_%'] = (shipmode_perf['delayed_orders'] / shipmode_perf['total_orders']) * 100
shipmode_perf
```

```python
category_analysis = df.groupby('category').agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    avg_profit_per_order=('profit', 'mean'),
    orders_count=('order_id', 'count')
).reset_index()
```

```python
category_analysis['profit_margin_%'] = (
    category_analysis['total_profit'] / category_analysis['total_sales']
) * 100
```

```python
category_analysis = category_analysis.sort_values(by='total_profit', ascending=False)
category_analysis
```

```python
subcat_analysis = df.groupby(['category','sub-category']).agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    avg_profit=('profit', 'mean')
).reset_index()
```

```python
subcat_analysis['is_profitable'] = subcat_analysis['total_profit'] > 0
```

```python
monthly_pattern = df.groupby('order_month').agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum')
).sort_values(by='total_sales', ascending=False)

monthly_pattern
```

```python
customer_orders = df.groupby('customer_id').agg(
    total_orders=('order_id', 'nunique'),
    total_spend=('sales', 'sum'),
    customer_profit=('profit', 'sum')
)

customer_orders['customer_type'] = np.where(
    customer_orders['total_orders'] > 5, 
    'Frequent Customer (>5 orders)', 
    'Occasional Customer (<=5 orders)'
)

customer_orders['customer_type'].value_counts()
```

```python
top_states = df.groupby('state')['profit'].sum().sort_values(ascending=False).head(5)
top_states
```

```python
bottom_states = df.groupby('state')['profit'].sum().sort_values().head(5)
bottom_states
```

```python

df['discount_range'] = np.where(
    df['discount'] == 0, '0% (No Discount)',
    np.where(
        df['discount'] <= 0.20, '1-20%',
        np.where(
            df['discount'] <= 0.50, '21-50%', '>50%'
        )
    )
)
```

```python
discount_impact = df.groupby('discount_range', observed=False).agg(
    orders_count=('order_id', 'count'),
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    avg_profit=('profit', 'mean')
)
discount_impact
```

```python
cat_summary = df.groupby('category')[['sales', 'profit', 'discount', 'quantity']].mean()
plt.figure(figsize=(8, 4))
sns.heatmap(cat_summary, annot=True, cmap='Blues', fmt=".2f")
plt.title("Average Metrics per Category", fontsize=12)
plt.ylabel("Category")
plt.xlabel("Metrics")
plt.show()
```
